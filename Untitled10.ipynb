```python
# Ячейка 3: Полный код ContentGPT Pro для Colab
import asyncio
import logging
from telegram import Update
from telegram.ext import Application, CommandHandler, MessageHandler, filters, ContextTypes
import google.generativeai as genai
from dotenv import load_dotenv
import os

# Настройка логирования
logging.basicConfig(
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    level=logging.INFO
)
logger = logging.getLogger(__name__)

# Загрузка конфигурации
load_dotenv()

TELEGRAM_TOKEN = os.getenv('TELEGRAM_BOT_TOKEN')
GEMINI_API_KEY = os.getenv('GEMINI_API_KEY')
ADMIN_ID = os.getenv('TELEGRAM_ADMIN_ID')
GEMINI_MODEL = os.getenv('GEMINI_MODEL', 'gemini-flash-latest')

# Настройка Gemini
genai.configure(api_key=GEMINI_API_KEY)
model = genai.GenerativeModel(GEMINI_MODEL)

# Обработчики команд
async def start(update: Update, context: ContextTypes.DEFAULT_TYPE):
    user = update.effective_user
    await update.message.reply_text(
        f"🎉 Привет, {user.first_name}!\n\n"
        f"Я ContentGPT Pro в Google Colab!\n"
        f"🤖 Модель: {GEMINI_MODEL}\n\n"
        f"📝 Команды:\n"
        f"/start - это сообщение\n"
        f"/help - справка\n"
        f"/generate тема - создать контент\n\n"
        f"Или просто напишите тему!"
    )

async def help_command(update: Update, context: ContextTypes.DEFAULT_TYPE):
    await update.message.reply_text(
        "📚 Справка ContentGPT Pro:\n\n"
        "1. Отправьте тему для генерации контента\n"
        "2. Используйте /generate тема\n"
        "3. Примеры тем:\n"
        "   • маркетинг в соцсетях\n"
        "   • искусственный интеллект\n"
        "   • создание контента\n\n"
        "⚠️ Бот работает в Google Colab облаке!"
    )

async def generate(update: Update, context: ContextTypes.DEFAULT_TYPE):
    if not context.args:
        await update.message.reply_text("❌ Укажите тему! Пример: /generate маркетинг")
        return

    topic = " ".join(context.args)
    msg = await update.message.reply_text(f"🧠 Генерирую: '{topic}'...")

    try:
        prompt = f"""Создай качественный контент на русском языке на тему: {topic}

        Формат:
        1. Заголовок
        2. Введение (2-3 предложения)
        3. Основная часть (3-5 пунктов)
        4. Заключение
        5. 3-5 хэштегов

        Стиль: профессиональный, но доступный."""

        response = model.generate_content(
            prompt,
            generation_config={"max_output_tokens": 800, "temperature": 0.7}
        )

        await msg.edit_text(f"✅ Готово!\n\n{response.text}")

    except Exception as e:
        await msg.edit_text(f"❌ Ошибка: {str(e)[:200]}")

async def handle_text(update: Update, context: ContextTypes.DEFAULT_TYPE):
    text = update.message.text
    if len(text) < 5:
        await update.message.reply_text("Напишите тему подробнее!")
        return

    # Если текст похож на тему, генерируем
    if len(text) < 100:
        context.args = [text]
        await generate(update, context)
    else:
        await update.message.reply_text("Отправляю на анализ...")
        response = model.generate_content(f"Проанализируй этот текст: {text}")
        await update.message.reply_text(f"📊 Анализ:\n{response.text[:1000]}")

async def error_handler(update: Update, context: ContextTypes.DEFAULT_TYPE):
    logger.error(f"Ошибка: {context.error}")
    if update and update.effective_message:
        await update.effective_message.reply_text("⚠️ Ошибка обработки запроса")

# Функция запуска бота
async def run_bot():
    print("🚀 Запуск бота в Colab...")

    # Создаем приложение
    app = Application.builder().token(TELEGRAM_TOKEN).build()

    # Регистрируем обработчики
    app.add_handler(CommandHandler("start", start))
    app.add_handler(CommandHandler("help", help_command))
    app.add_handler(CommandHandler("generate", generate))
    app.add_handler(MessageHandler(filters.TEXT & ~filters.COMMAND, handle_text))
    app.add_error_handler(error_handler)

    print(f"✅ Бот запущен! Токен: {TELEGRAM_TOKEN[:10]}...")
    print("📱 Ищите бота в Telegram и отправьте /start")
    print("🔄 Бот работает...")

    # Запускаем опрос
    await app.run_polling(allowed_updates=Update.ALL_TYPES)
```

```python
# Ячейка 4: Запуск бота (основная ячейка)
import nest_asyncio
nest_asyncio.apply()  # Для работы asyncio в Colab

print("=" * 50)
print("🤖 CONTENTGPT PRO В GOOGLE COLAB")
print("=" * 50)

# Проверка ключей
if not TELEGRAM_TOKEN or TELEGRAM_TOKEN.startswith('ВАШ_'):
    print("❌ Ошибка: Telegram токен не настроен!")
    print("📝 Получите токен у @BotFather")
elif not GEMINI_API_KEY or GEMINI_API_KEY.startswith('ВАШ_'):
    print("❌ Ошибка: Gemini API ключ не настроен!")
    print("📝 Получите ключ: https://makersuite.google.com/app/apikey")
else:
    print("✅ Ключи настроены")
    print(f"🤖 Модель: {GEMINI_MODEL}")
    print(f"👑 Админ ID: {ADMIN_ID}")
    print("\n" + "=" * 50)

    try:
        # Запускаем бота
        await run_bot()
    except RuntimeError:
        # Если await не работает на верхнем уровне
        print("\n🔄 Запускаю через asyncio.run()...")
        import asyncio
        asyncio.run(run_bot())
```

```python
# Ячейка 5: Бот с веб-интерфейсом в Colab
from IPython.display import display, HTML
import ipywidgets as widgets

# Создаем виджеты
telegram_token_input = widgets.Password(
    description="Telegram Token:",
    placeholder='токен от @BotFather'
)

gemini_key_input = widgets.Password(
    description="Gemini Key:",
    placeholder='ключ от Google AI'
)

admin_id_input = widgets.Text(
    description="Ваш ID:",
    placeholder='ID из @userinfobot'
)

start_button = widgets.Button(
    description="🚀 Запустить бота",
    button_style='success',
    icon='play'
)

output = widgets.Output()

def on_start_button_clicked(b):
    with output:
        output.clear_output()

        # Получаем значения
        TELEGRAM_TOKEN = telegram_token_input.value
        GEMINI_API_KEY = gemini_key_input.value
        ADMIN_ID = admin_id_input.value

        if not all([TELEGRAM_TOKEN, GEMINI_API_KEY, ADMIN_ID]):
            print("❌ Заполните все поля!")
            return

        # Создаем .env
        with open('.env', 'w') as f:
            f.write(f"""TELEGRAM_BOT_TOKEN={TELEGRAM_TOKEN}
GEMINI_API_KEY={GEMINI_API_KEY}
TELEGRAM_ADMIN_ID={ADMIN_ID}
GEMINI_MODEL=gemini-flash-latest
""")

        print("✅ Конфигурация сохранена")
        print("🔄 Запускаю бота...")

        # Здесь можно запустить бота в фоне
        print(f"🤖 Бот запущен с токеном: {TELEGRAM_TOKEN[:10]}...")
        print("📱 Перейдите в Telegram и отправьте /start боту")
        print("⚠️ Для остановки: Kernel → Interrupt Execution")

start_button.on_click(on_start_button_clicked)

# Отображаем интерфейс
display(HTML("<h3>🎛️ Панель управления ContentGPT Pro</h3>"))
display(telegram_token_input)
display(gemini_key_input)
display(admin_id_input)
display(start_button)
display(output)
```
